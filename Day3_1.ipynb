```python
import numpy as np
from numpy.random import randint as ri
import pandas as pd

# Generate 64 random integers from 1 to 6
# Reshape them into an 8x8 NumPy array
tb = np.array(ri(1, 7, 64)).reshape(8, 8)
print(tb)

# Define column names
col = ['X', 'Y', 'Z', 'N', 'M', 'O', 'P', 'R']

# Define row/index names
row = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H']

# Create DataFrame using custom row and column labels
df = pd.DataFrame(
    data=tb,
    index=row,
    columns=col
)

print(df)
print("-" * 50)


# Drop row 'A'
# By default, drop() removes rows because axis=0
# This does not permanently modify df unless inplace=True is used
print(df.drop('A'))

# To permanently remove row 'A':
# df.drop('A', axis=0, inplace=True)

# To remove column 'O':
# df.drop('O', axis=1, inplace=True)

print("-" * 50)


# Select rows using integer positions
# iloc[0:3] selects rows at positions 0, 1, and 2
print(df.iloc[0:3])
print("-" * 50)


# Select all rows and column at position 2
# Column position 2 is 'Z'
print(df.iloc[:, 2])
print("-" * 50)


# Select the value at row position 1 and column position 2
print(df.iloc[1, 2])
print("-" * 50)


# Select specific rows and columns by integer positions
# Rows: positions 1 and 3
# Columns: positions 3 and 6
print(df.iloc[[1, 3], [3, 6]])
print("-" * 50)


# Multiply every numerical value in the DataFrame by 6
df = df * 6

# Create a Boolean DataFrame
# True where the value is greater than 20
print(df > 20)

# Store the Boolean condition in bdf
bdf = df > 31

# Display only values where condition is True
# Other values become NaN
print(df[bdf])
print("-" * 50)

print(bdf)
print("-" * 50)

print(df)


# Correct filtering:
# 'A' is a row label, not a column name.
# Use row filtering with index labels through loc.

# Select row 'A' and check values greater than 15
print(df.loc["A"] > 15)

# Select rows where column 'O' is greater than 89
print(df[df["O"] > 89])

# Select rows where column 'O' is greater than 89
# and row index is not directly used as a column condition
```

```python
import pandas as pd

students = pd.DataFrame({
    "StudentID": [101, 102, 103, 104, 105],
    "Name": ["Aman", "Riya", "Karan", "Simran", "Arjun"],
    "Department": ["CSE", "IT", "CSE", "ECE", "IT"]
})

marks = pd.DataFrame({
    "StudentID": [101, 102, 103, 106, 107],
    "Marks": [85, 92, 78, 88, 95],
    "Grade": ["A", "A+", "B", "A", "A+"]
})

r=pd.merge(students,marks,how='inner')
print(r)
rl=pd.merge(students,marks,on="StudentID",how='left')
print(rl)
rr=pd.merge(students,marks,on="StudentID",how='right')
print(rr)
ro=pd.merge(students,marks,on="StudentID",how='outer')
print(ro)
```
